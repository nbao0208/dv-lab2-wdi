## 1. Xử lý dữ liệu

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display, Markdown


# Đọc file CSV
file_path = "../../data/wdi_cleaned.csv"
df = pd.read_csv(file_path)

df = df.rename(columns={
    'Country Name': 'country',
    'Indicator Name': 'indicator',
    'Year': 'year',
    'Value': 'value'
})

# Xem dữ liệu
df.info()
df['indicator'].unique()
```

## 2. Phân tích

```python
display(Markdown("# **Chương 3: Cái giá của sự phát triển & Công nghệ**"))


# CÂU HỎI 7: Dấu chân Carbon (CO2)
display(Markdown("## 📖 7. Dấu chân Carbon (CO2): Ai đang làm nóng trái đất?"))
display(Markdown("""
**Câu hỏi: Top 20 quốc gia nào đang xả thải khí nhà kính (bình quân đầu người) cao nhất thế giới?**
* **Chỉ số:** `CO2 emissions (metric tons per capita)`.
* **Biểu đồ:** Horizontal Bar Chart (Biểu đồ cột ngang).
* **Mục tiêu:** Hiển thị top 20 quốc gia có lượng phát thải CO2 bình quân đầu người cao nhất trong năm gần nhất có dữ liệu, 
                 giúp chúng ta nhận diện những "thủ phạm" chính trong việc làm nóng trái đất.
"""))

# Biểu đồ 7: Top 20 quốc gia có lượng phát thải CO2 bình quân đầu người cao nhất
co2 = df[df['indicator'].str.contains("CO2 emissions", case=False, na=False)]
# Lấy năm mới nhất
latest_year = co2['year'].max()
co2_latest = co2[co2['year'] == latest_year]

# Top 20
top20 = co2_latest.sort_values(by='value', ascending=False).head(20)

plt.figure(figsize=(10,6))
sns.barplot(data=top20, x='value', y='country')

plt.title(f"Top 20 CO2 Emitters per Capita ({latest_year})")
plt.xlabel("CO2 (tấn/người)")
plt.ylabel("Quốc gia")

plt.show()

display(Markdown("""
<div style="text-align: justify;">
                 
**Phân tích:**
* **Nhóm thải nhiều nhất:** Dựa trên biểu đồ, Malawi là nước dẫn đầu với con số xấp xỉ 34 tấn / người, theo sau là
                 Nigeria (khoảng 29 tấn / người).
* **Nhóm trung bình**: Các quốc gia như Guinea, Somalia và Liberia dao động trong khoảng từ 14 đến 17 tấn/người.
* **Nhóm thải ít nhất:** Các nước như Kuwait, Guinea có lượng phát thải CO2 bình quân đầu người rất thấp, chỉ khoảng 0.1 đến 0.5 tấn/người.
* **Nhận xét:** Sự chênh lệch lớn giữa các quốc gia có thể cho thấy rằng một số nước đang gánh chịu hậu quả nặng nề của biến đổi khí hậu dù họ không phải là nguồn thải CO2 chính, 
                trong khi những nước thải nhiều lại có thể đang hưởng lợi từ sự phát triển kinh tế mà không phải chịu nhiều tác động trực tiếp.

**Đáng lưu ý:** Các quốc gia vốn có nền công nghiệp dầu mỏ hoặc thu nhập cao như
                Kuwait lại có lượng phát thải CO2 bình quân đầu người rất thấp. Kết quả từ biểu đồ này đang cho thấy một xu hướng ngược hoàn toàn so với dữ liệu 
                lịch sử của World Bank (thông thường các nước phát triển và vùng Vịnh mới là nhóm phát thải cao nhất).
                 
**Các khả năng:** 
* **Sự dịch chuyển của chuỗi cung ứng**: Các ngành công nghiệp nặng đang dịch chuyển 
                mạnh mẽ sang khu vực Châu Phi khiến lượng thải CO2 tại đây tăng vọt.
* **Dữ liệu không đầy đủ hoặc lỗi**: Có thể có vấn đề về chất lượng dữ liệu, 
                hoặc có sự thay đổi trong cách tính toán phát thải CO2.
"""))


# CÂU HỎI 8: Nỗ lực xanh hóa (Energy Transition)
display(Markdown("## 📖 8. Nỗ lực xanh hóa (Energy Transition)"))
display(Markdown("""
**Câu hỏi: Quá trình chuyển đổi sang năng lượng tái tạo diễn ra nhanh đến mức nào ở các nhóm thu nhập khác nhau?**
* **Chỉ số:** `Renewable energy consumption (% of total final energy consumption)`.
* **Biểu đồ:** Line Chart (Biểu đồ đường).
* **Mục tiêu:** Hiển thị xu hướng sử dụng năng lượng tái tạo theo nhóm thu nhập qua các năm,
                    giúp chúng ta quan sát quá trình, tốc độ chuyển đổi năng lượng xanh của từng nhóm thu nhập,
                từ đó nhận diện những "điểm sáng" trong nỗ lực xanh hóa toàn cầu.
"""))

# Biểu đồ 8: Xu hướng sử dụng năng lượng tái tạo theo Nhóm thu nhập
energy_indicator = 'Renewable energy consumption (% of total final energy consumption)'
df_energy = df[(df['indicator'] == energy_indicator) & 
                     (df['Income Group'].notna())]

# Tính trung bình theo Nhóm thu nhập mỗi năm
df_energy_trend = df_energy.groupby(['year', 'Income Group'])['value'].mean().reset_index()

plt.figure(figsize=(12, 6))
sns.lineplot(data=df_energy_trend, x='year', y='value', hue='Income Group')
plt.title('Xu hướng sử dụng năng lượng tái tạo theo nhóm thu nhập')
plt.ylabel('% tổng năng lượng tiêu thụ')
plt.xlabel('Năm')
plt.show()

display(Markdown("""
<div style="text-align: justify;">
                 
**Phân tích:**
* **Nhóm thu nhập thấp (Low income)**: Nhóm này có xu hướng sử dụng năng lượng tái tạo cao nhất.
* **Nhóm thu nhập trung bình thấp (Lower middle income)**: Nhóm này có mức sử dụng năng lượng tái tạo trung bình.
* **Nhóm thu nhập trung bình cao (Upper middle income)**: Nhóm này có mức sử dụng năng lượng tái tạo ở mức thấp.
* **Nhóm thu nhập cao (High income)**: Nhóm này có xu hướng sử dụng năng lượng tái tạo thấp nhất.
                 
**Nhận xét:**
* Ở các nước nghèo, phần lớn năng lượng tái tạo 
    đến từ việc đốt củi, than củi, chất thải nông nghiệp để đun nấu. 
    Điều này giải thích tại sao tỷ lệ của họ rất cao nhưng đời sống và công nghệ lại chưa phát triển.
* Các quốc gia thu nhập cao có thể đang sử dụng nhiều năng lượng hóa thạch do sự phát 
    triển công nghiệp mạnh mẽ và nhu cầu năng lượng lớn, dù họ có khả năng tài chính để chuyển 
    đổi sang năng lượng tái tạo. 
* Sự chênh lệch này có thể cho thấy rằng các quốc gia giàu có đang không tận dụng tối đa tiềm năng của năng lượng tái tạo,
    trong khi các quốc gia nghèo hơn lại đang dẫn đầu trong việc sử dụng năng lượng xanh, có thể 
    do họ không có nhiều lựa chọn khác.

**Điểm lưu ý:** Sự biến động cực mạnh ở năm 2022, nhóm thu nhập thấp tăng vọt lên gần 100%, 
                 trong khi nhóm thu nhập cao lại lao dốc xuống dưới 20%.

**Nguyên do:** Điểm "gãy" này khả năng cao là do **thiếu hụt dữ liệu** khiến con số trung bình bị 
                 sai lệch, chứ không phải một sự thay đổi đột ngột của nền kinh tế toàn cầu trong vòng 1 năm.
                 
"""))

# CÂU HỎI 9: Kỷ nguyên kết nối toàn cầu
display(Markdown("## 📖 9. Kỷ nguyên kết nối toàn cầu"))
display(Markdown("""
**Câu hỏi: Mạng Internet đã phủ sóng thế giới với tốc độ hàm mũ như thế nào kể từ đầu những năm 2000?**
* **Chỉ số:** `Individuals using the Internet (% of population)`.
* **Biểu đồ:** S-curve Line Chart (đường cong chữ S).
* **Mục tiêu:** Hiển thị xu hướng tăng trưởng số lượng người dùng Internet qua các năm, 
                 giúp chúng ta nhận diện những "điểm sáng" trong quá trình kết nối toàn cầu.
"""))

# Biểu đồ 9: Xu hướng sử dụng Internet theo nhóm thu nhập
# 1. Lọc dữ liệu cho chỉ số Internet
internet_indicator = 'Individuals using the Internet (% of population)'
df_internet = df[(df['indicator'] == internet_indicator) & 
                       (df['Income Group'].notna())]

# 2. Tính trung bình theo từng năm và Nhóm thu nhập
df_internet_trend = df_internet.groupby(['year', 'Income Group'])['value'].mean().reset_index()
# 3. Vẽ biểu đồ
plt.figure(figsize=(12, 6))
sns.lineplot(data=df_internet_trend, x='year', y='value', hue='Income Group', marker='o', markersize=4)

plt.title("Sự bùng nổ Internet toàn cầu theo Nhóm thu nhập (1990 - 2022)")
plt.xlabel("Năm")
plt.ylabel("% Dân số sử dụng Internet")
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend(title='Nhóm thu nhập', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()
plt.show()

display(Markdown("""
<div style="text-align: justify;">
                 
**Phân tích:**
* **Mô hình S-Curve:** Biểu đồ thể hiện sự chuyển dịch hoàn hảo từ giai đoạn **Innovation** 
                 (đổi mới - những năm 1990) sang giai đoạn **Mass Adoption** (áp dụng đại trà - sau năm 2005). 
* **Điểm bùng phát:** Đối với nhóm **High income**, điểm bùng phát diễn ra sớm vào năm 2000.
                 Trong khi đó, nhóm **Upper middle income** chứng kiến sự tăng trưởng dốc đứng từ sau năm 2005, 
                 tương ứng với cuộc cách mạng Smartphone và hạ tầng mạng 3G/4G toàn cầu.           
* **Phân hóa kĩ thuật số:**
    * **Nhóm Thu nhập Cao:** Đạt ngưỡng bão hòa từ năm 2018 với tỷ lệ trên 90%. 
                 Tại đây, Internet không còn là công nghệ mới mà đã trở thành nền tảng hạ tầng bắt buộc.
    * **Sự trỗi dậy của Nhóm Trung bình:** Đây là phần thú vị nhất. Tốc độ dốc của đường 
                 màu Đỏ cho thấy các nước này đã "nhảy cóc" (leapfrogging) qua giai đoạn PC để tiến thẳng lên Internet di động. 
                 Việc đạt tỷ lệ 70-80% mở ra cơ hội khổng lồ cho các ứng dụng **Fintech** và **E-commerce**. 
    * **Nhóm Thu nhập Thấp:** Dù có tăng trưởng nhưng vẫn bị bỏ lại phía sau với tỷ lệ dưới 25%. Sự chênh lệch 
                 này giải thích tại sao nhiều giải pháp công nghệ hiện đại vẫn khó tiếp cận được các 
                 khu vực này do thiếu hụt hạ tầng mạng lõi.
**Điểm lưu ý:** Đường **Low income** có dấu hiệu rơi tự do về gần 0% vào cuối năm 2024.
                 
**Nguyên do:** Rất có thể đây là một lỗi dữ liệu hoặc thiếu hụt dữ liệu nghiêm trọng, chứ không phải một sự sụp 
                đổ thực sự của kết nối Internet toàn cầu.
"""))
```
